## Question 2 - Research idea number two

Propose a second idea that tests a meaningfully different hypothesis.

### Answer

**Market and scope:**

I would study pairs of similar US companies, such as two firms in the same industry, over a one-to-three-month holding period.

**Investment thesis:**

Sometimes two similar companies move apart even though their businesses are still related. I would test whether the cheaper company tends to catch up with the more expensive one.

**Potential edge:**

This is a relative-value idea, so it does not depend completely on the market going up. However, the relationship could disappear after a major business or regulatory change.

**Data and sources:**

I would use adjusted prices, dividends, company filings, sector classifications, earnings dates, and a market benchmark. I would check that the companies are still comparable.

**Research approach:**

I would calculate the normal price relationship using older data. When the spread becomes unusually large, I would buy the cheaper stock and short the expensive one, then exit when the spread returns closer to normal.

**Validation and risks:**

I would compare it with a long-only portfolio and a sector benchmark. I would include commissions, spreads, slippage, borrow fees, and dividends on the short position. I would reject the idea if the result depends on only one pair or one historical crisis.
